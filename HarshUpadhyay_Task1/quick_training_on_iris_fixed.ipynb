```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

# Iris Flower Classification

OIBSIP Data Science Internship - Task 1

```python
data = pd.read_excel("iris-train.xlsx")
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data.describe()
```

```python
sns.countplot(x=data.iloc[:,-1])
plt.title('Distribution of Iris Species')
plt.show()
```

```python
x=data.iloc[:,:-1]
y=data.iloc[:,-1]

X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
prediction = model.predict([[5.4,3.0,4.5,1.5]])
print("Predicted Species:", prediction[0])
```

# Conclusion

A Random Forest Classification model was developed to classify Iris flowers into different species. The model achieved high accuracy and successfully predicted the species of Iris flowers using their measurements.
